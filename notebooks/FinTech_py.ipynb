```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_excel("default of credit card clients.xlsx")
df()
```

```python
#add new column for credit utilization ratio
df["credit utilization ratio"]=df["BILL_AMT1"]/df["LIMIT_BAL"]
df
```

```python
#payment to bill ratio
df['pay_to_bill_ratio'] = df['PAY_AMT1'] / (df['BILL_AMT1'] + 1)
df
```

```python
#mean monthly bill
df['mean_monthly_bill'] = (df['BILL_AMT1'] + df['BILL_AMT2'] + df['BILL_AMT3'] + df['BILL_AMT4'] + df['BILL_AMT5'] + df['BILL_AMT6']) / 6
df
```

```python
#mean monthly payment
df['mean_monthly_payment'] = (df['PAY_AMT1'] + df['PAY_AMT2'] + df['PAY_AMT3'] + df['PAY_AMT4'] + df['PAY_AMT5'] + df['PAY_AMT6']) / 6
df
```

```python
#max payment delay
df['max_payment_delay'] = df[['PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']].max(axis=1)
df
```
